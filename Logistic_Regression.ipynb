**Data Cleaning**

```python
df.isnull().sum().to_frame().style.set_properties(
    **{
    'background-color' : 'purple',
    'color':'white',
    'border-color':'white',
    'border-style':'solid'

}
)
```

```python
col_list = df.columns
```

```python
for x in col_list:
  if df[x].dtypes =='object':
    df[x] = df[x].fillna(df[x].mode()[0])
  else:
    df[x] = df[x].fillna(df[x].mean())
```

```python
df.isnull().sum().to_frame().style.set_properties(
    **{
    'background-color' : 'purple',
    'color':'white',
    'border-color':'white',
    'border-style':'solid'

}
)
```

```python
df.drop_duplicates(inplace=True)
```

```python
for col in df.columns:
    if (df[col].dtype != 'object'):
        sns.boxplot(df[col])
        plt.xlabel(col)
        plt.show()
```

```python
oultier_col = ['age', 'duration','campaign','cons.conf.idx']
```

```python
for x in oultier_col:
  Q1 = df[x].quantile(0.25)
  Q3 = df[x].quantile(0.75)
  IQR = Q3-Q1

  LL =    Q1-1.5*IQR
  UU =    Q3+1.5*IQR

  df = df[(df[x] >= LL) & (df[x] <=UU)]

```

```python
for col in df.columns:
    if (df[col].dtype != 'object'):
        sns.boxplot(df[col])
        plt.xlabel(col)
        plt.show()
```

```python
x = df.iloc[: ,:-1]

y = df.iloc[:, -1]
```

```python
from sklearn.feature_selection import RFE

from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

```python
rfe = RFE(model)
```

```python
rfe.fit(x, y)
```

```python
rfe.support_
```

```python
selected_indices  = [ i for i , data in enumerate(rfe.support_)if data == True]
```
